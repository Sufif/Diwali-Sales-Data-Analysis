Exploratary Data Analysis (EDA)

```python
df.columns
```

```python
a = sns.countplot(x = "Gender", data = df)

for bars in a.containers:
    a.bar_label(bars)
```

```python
b = df.groupby (["Gender"],as_index=False) ["Amount"].sum().sort_values(by="Amount",ascending=False)
b
```

```python
sns.barplot(x = "Gender" ,y = "Amount",data = b)


```

```python
c = sns.countplot(data = df,x = "Age Group",hue = "Gender")

for bars in c.containers:
    c.bar_label(bars)
```

```python
sales_age = df.groupby(["Age Group"],as_index=False)["Amount"].sum().sort_values(by="Amount",ascending=False)
sns.barplot(x="Age Group",y = "Amount",data = sales_age)
```

```python
sales_age = df.groupby("Age Group",as_index=False)["Amount"].sum()
print(sales_age)
```

```python
#plt.figure(figsize=(8,5))

sns.barplot(x="Age Group",y="Amount",data=sales_age)
#plt.show()
```

```python
a = sns.countplot(data = df,x = "Age Group", hue = "Gender")
```

```python
a = sns.countplot(data = df,x = "Age Group", hue = "Gender")

for bars in a.containers:
    a.bar_label(bars)
```

```python
sales_state = (
    df.groupby("State", as_index=False)["Orders"].sum().sort_values(by="Orders", ascending=False).head(7))

sns.set(rc={"figure.figsize": (15, 7)})

b = sns.barplot(
    data=sales_state,
    x="State",
    y="Orders",
    order=sales_state["State"]
)

b.bar_label(b.containers[0])
```

```python
sales_state = ( df.groupby("State", as_index=False)["Amount"].sum().sort_values(by="Amount", ascending=False).head(7))

a = sns.barplot(data = sales_state,x = "State", y = "Amount")

a.bar_label(a.containers[0])
```

```python
a = sns.countplot(data = df,x = "Marital_Status")


sns.set(rc={"figure.figsize":(6,3)})
a.bar_label(a.containers[0])
```

```python
a = sns.countplot(data = df,x = "Occupation")

sns.set(rc={"figure.figsize":(20,10)})

for bars in a.containers:
    a.bar_label(bars)
```

```python
sales = df.groupby(["Occupation"],as_index=False)["Amount"].sum().sort_values(by="Amount",ascending=False)

sns.set(rc={"figure.figsize":(20,5)})

b = sns.barplot(data = sales,x = "Occupation",y = "Amount")

for bars in b.containers:
    b.bar_label(bars)
```
